 1: Load the Dataset




# New Section

```python
import pandas as pd

# Load the dirty dataset
df = pd.read_csv('movie_dataset_dirty.csv', on_bad_lines='skip', engine='python')
# Show initial data
print("Initial shape:", df.shape)
df.head()
```

2: Remove Duplicates


```python
# Remove duplicate rows
df = df.drop_duplicates()

print("After removing duplicates:", df.shape)
df.head()
```

3: Strip Whitespace & Normalize Text

```python
# Strip whitespaces and convert to title case
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].astype(str).str.strip().str.title()

df.head()
```

Cell 4: Handle Missing Values

```python
# Drop rows with any missing values
df = df.dropna()

print("After dropping missing values:", df.shape)
df.head()
```

5: Save the Cleaned Dataset

```python
# Save the cleaned dataset
df.to_csv('movie_dataset_cleaned.csv', index=False)
print("Cleaned dataset saved as 'movie_dataset_cleaned.csv'")
```
